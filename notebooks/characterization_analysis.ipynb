# Ag@QNPs Characterization Analysis

This notebook visualizes physicochemical characterization data for silver-quercetin nanoparticles (Ag@QNPs) reported in:

> Asadi, S. et al. *Enhanced urease inhibitory activity of quercetin via conjugation with silver nanoparticles: synthesis, characterization, and DFT study.* Scientific Reports 15, 11892 (2025). https://doi.org/10.1038/s41598-025-96684-2

**Data source:** All values are from the published paper. This notebook was written to document and visualize the experimental results in a reproducible format.

---

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Color palette
COLOR_AG   = '#7393B3'   # steel blue for AgNPs
COLOR_AGQN = '#C07B3A'   # amber for Ag@QNPs
COLOR_Q    = '#6B8E6B'   # muted green for Quercetin
COLOR_HU   = '#A0706E'   # muted red for Hydroxyurea

print('Libraries loaded.')
```

## 1. UV-Vis Absorption Peaks

Surface plasmon resonance (SPR) peak shift from 407 nm (AgNPs) to 445 nm (Ag@QNPs) confirms quercetin loading.

```python
fig, ax = plt.subplots(figsize=(6, 4))

compounds = ['AgNPs', 'Ag@QNPs']
peaks     = [407, 445]
colors    = [COLOR_AG, COLOR_AGQN]

bars = ax.bar(compounds, peaks, color=colors, width=0.4, edgecolor='white', linewidth=1.2)

for bar, val in zip(bars, peaks):
    ax.text(bar.get_x() + bar.get_width()/2, val + 3,
            f'{val} nm', ha='center', va='bottom', fontsize=11, fontweight='bold')

ax.annotate('', xy=(1, 445), xytext=(0, 407),
            arrowprops=dict(arrowstyle='->', color='gray', lw=1.5))
ax.text(0.5, 426, '+38 nm shift\n(quercetin loading)', ha='center', va='center',
        fontsize=9, color='gray', style='italic')

ax.set_ylabel('SPR Peak Wavelength (nm)', fontsize=11)
ax.set_title('UV-Vis Absorption Peaks', fontsize=13, fontweight='bold')
ax.set_ylim(380, 470)
ax.spines[['top','right']].set_visible(False)

plt.tight_layout()
plt.savefig('../figures/uvvis_peaks.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved: figures/uvvis_peaks.png')
```

## 2. Hydrodynamic Diameter (DLS)

Size increase from AgNPs to Ag@QNPs confirms successful quercetin coating.

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# --- DLS Size ---
sizes  = [70.85, 104.5]
errors = [9.68, 8.53]

axes[0].bar(compounds, sizes, yerr=errors, color=colors, width=0.4,
            capsize=6, edgecolor='white', linewidth=1.2)
for i, (s, e) in enumerate(zip(sizes, errors)):
    axes[0].text(i, s + e + 2, f'{s} ± {e}', ha='center', fontsize=9, fontweight='bold')

axes[0].set_ylabel('Hydrodynamic Diameter (nm)', fontsize=11)
axes[0].set_title('DLS Particle Size', fontsize=13, fontweight='bold')
axes[0].set_ylim(0, 135)
axes[0].spines[['top','right']].set_visible(False)

# --- Zeta Potential ---
zp     = [-23.5, -15.2]
zp_err = [3.16, 2.3]

axes[1].bar(compounds, zp, yerr=zp_err, color=colors, width=0.4,
            capsize=6, edgecolor='white', linewidth=1.2)
for i, (z, e) in enumerate(zip(zp, zp_err)):
    axes[1].text(i, z - e - 1.5, f'{z} ± {e} mV', ha='center', fontsize=9, fontweight='bold')

axes[1].axhline(-15, color='gray', linestyle='--', linewidth=0.8, alpha=0.6)
axes[1].text(1.25, -14.5, 'stability\nthreshold', fontsize=8, color='gray')
axes[1].set_ylabel('Zeta Potential (mV)', fontsize=11)
axes[1].set_title('Zeta Potential', fontsize=13, fontweight='bold')
axes[1].set_ylim(-32, 2)
axes[1].spines[['top','right']].set_visible(False)

plt.suptitle('DLS Characterization — AgNPs vs Ag@QNPs', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('../figures/dls_zeta.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved: figures/dls_zeta.png')
```

## 3. XRD Peak Positions

Peaks at 2θ = 38.10°, 44.32°, 64.44°, 77.41° confirm face-centered cubic (FCC) silver crystal structure (JCPDS 087-0717).
Crystallite size calculated via Debye-Scherrer equation: **D = 0.62 nm**.

```python
xrd = pd.read_csv('../data/xrd_peaks.csv')

fig, ax = plt.subplots(figsize=(7, 4))

# Simulate stick pattern
relative_intensity = [100, 45, 30, 20]   # approximate relative intensities
for i, (row, intensity) in enumerate(zip(xrd.itertuples(), relative_intensity)):
    ax.vlines(row.two_theta_deg, 0, intensity, color=COLOR_AG, linewidth=3)
    ax.text(row.two_theta_deg, intensity + 2,
            f'{row.two_theta_deg}°\n({row.hkl})',
            ha='center', fontsize=9, fontweight='bold')

ax.set_xlabel('2θ (degrees)', fontsize=11)
ax.set_ylabel('Relative Intensity (%)', fontsize=11)
ax.set_title('XRD Peak Positions — AgNPs\n(FCC Ag, JCPDS 087-0717)', fontsize=13, fontweight='bold')
ax.set_xlim(30, 90)
ax.set_ylim(0, 125)
ax.spines[['top','right']].set_visible(False)

ax.text(0.97, 0.97, 'Crystallite size (Scherrer): 0.62 nm',
        transform=ax.transAxes, ha='right', va='top',
        fontsize=9, color='gray', style='italic')

plt.tight_layout()
plt.savefig('../figures/xrd_peaks.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved: figures/xrd_peaks.png')
```

## 4. Urease Inhibitory Activity — IC₅₀ Comparison

Ag@QNPs show ~20× potency improvement over free quercetin (IC₅₀ basis on quercetin-equivalent concentration).

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

# --- IC50 bar chart ---
ic50_compounds = ['Quercetin\n(free)', 'Ag@QNPs\n(quercetin eq.)', 'Hydroxyurea\n(positive control)']
ic50_vals      = [9.581, 0.473, 7.6]
ic50_errs      = [0.38, 0.09, 0]
ic50_colors    = [COLOR_Q, COLOR_AGQN, COLOR_HU]

bars = axes[0].bar(ic50_compounds, ic50_vals, yerr=ic50_errs,
                   color=ic50_colors, width=0.5, capsize=6,
                   edgecolor='white', linewidth=1.2)
for bar, val, err in zip(bars, ic50_vals, ic50_errs):
    label = f'{val} ± {err}' if err > 0 else f'{val}'
    axes[0].text(bar.get_x() + bar.get_width()/2, val + err + 0.15,
                 label, ha='center', fontsize=8.5, fontweight='bold')

axes[0].set_ylabel('IC₅₀ (µg/mL)', fontsize=11)
axes[0].set_title('IC₅₀ Comparison', fontsize=13, fontweight='bold')
axes[0].set_ylim(0, 13)
axes[0].spines[['top','right']].set_visible(False)

# --- Potency fold change (log scale) ---
fold_labels = ['Quercetin\nvs Ag@QNPs', 'Ag@QNPs\nvs Hydroxyurea']
fold_vals   = [9.581 / 0.473, 7.6 / 0.473]
fold_colors = [COLOR_AGQN, COLOR_HU]

axes[1].barh(fold_labels, fold_vals, color=fold_colors, edgecolor='white', linewidth=1.2)
for i, v in enumerate(fold_vals):
    axes[1].text(v + 0.3, i, f'{v:.1f}×', va='center', fontsize=10, fontweight='bold')

axes[1].set_xlabel('Fold improvement in potency', fontsize=11)
axes[1].set_title('Relative Potency of Ag@QNPs', fontsize=13, fontweight='bold')
axes[1].set_xlim(0, 25)
axes[1].spines[['top','right']].set_visible(False)

plt.suptitle('Urease Inhibitory Activity', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('../figures/ic50_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved: figures/ic50_comparison.png')
```
